```python
#Adding Year! 
from pathlib import Path
import pandas as pd

RAW_FILE = Path("/Users/pranavongole/Documents/DataForge365/PROJECT 00/US-Immigration-Data/data/raw/state_dept_niv_detail_fy97-24.xlsx")
OUTPUT_FILE = Path("/Users/pranavongole/Documents/DataForge365/PROJECT 00/US-Immigration-Data/data/processed/visa_by_year.csv")  # adjust as needed

def sheet_name_to_year(sheet_name: str) -> int:
    suffix = int(sheet_name.replace("FY", ""))
    return 1900 + suffix if suffix >= 97 else 2000 + suffix

merged_frames = []
with pd.ExcelFile(RAW_FILE) as workbook:
    for sheet in workbook.sheet_names:
        df = pd.read_excel(workbook, sheet_name=sheet)
        df["Year"] = sheet_name_to_year(sheet)
        merged_frames.append(df)

merged = pd.concat(merged_frames, ignore_index=True)
merged.to_csv(OUTPUT_FILE, index=False)
```
